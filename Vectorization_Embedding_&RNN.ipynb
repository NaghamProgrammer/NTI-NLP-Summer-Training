#Part 1

```python
texts = [
    "AI is changing the world",
    "Machine learning is amazing",
    "Deep learning powers many applications",
    "I enjoy learning NLP",
    "NLP is fun and useful"
]

sentences = [text.lower().split() for text in texts]
```

##Task 1
Train a Word2Vec model using vector_size=30, window=2, min_count=1.


```python
w2v_model = Word2Vec(
    sentences=sentences,
    vector_size=30,
    window=2,
    min_count=1,
    workers=4,
    seed=42
)
```

##Task 2
Find the number of unique words and identify the most frequent word without printing the entire vocabulary.


```python
word_counts = Counter()

for sentence in sentences:
    word_counts.update(sentence)

print("Number of unique words:", len(word_counts))

most_common = word_counts.most_common(1)[0]

print("Most frequent word:", most_common[0])
print("Frequency:", most_common[1])
```

##Task 3
Print the word vectors of any two words of your choice.


```python
print("Vector for 'learning':")
print(w2v_model.wv["learning"])

print("\nVector for 'applications':")
print(w2v_model.wv["applications"])
```

##Task 4
Find the two most similar words to "learning".

```python
similar = w2v_model.wv.most_similar("learning", topn=2)

print("Two most similar words to 'learning':")

for word, score in similar:
    print(f"{word}: {score:.4f}")
```

##Task 5
Choose any two pairs of words, calculate their similarity, and briefly explain the results.


```python
similarity = w2v_model.wv.similarity("machine", "learning")

print(f"Similarity(machine, learning) = {similarity:.4f}")
```

#Part 2

```python
texts = [
    "I love AI",
    "AI is amazing",
    "I hate bugs",
    "Coding is fun",
    "This movie is terrible",
    "This movie is great"
]

labels = [1, 1, 0, 1, 0, 1]
```

##Task 1
Convert the text into sequences using TensorFlow Tokenizer and print only the sequences.


```python
tokenizer = Tokenizer()

tokenizer.fit_on_texts(texts)

sequences = tokenizer.texts_to_sequences(texts)

print("Sequences:")
print(sequences)
```

##Task 2
Pad all sequences to the maximum sentence length in the dataset.

```python
max_length = max(len(seq) for seq in sequences)

padded_sequences = pad_sequences(
    sequences,
    maxlen=max_length,
    padding="post"
)

print(padded_sequences)
```

##Task 3
Build a model with Embedding, SimpleRNN, and Dense output layers. Choose the embedding dimension and the number of RNN units yourself.


```python
vocab_size = len(tokenizer.word_index) + 1

embedding_dim = 16
rnn_units = 16

model = Sequential([
    Embedding(
        input_dim=vocab_size,
        output_dim=embedding_dim,
        input_length=max_length
    ),
    SimpleRNN(rnn_units),
    Dense(1, activation="sigmoid")
])

model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

model.summary()
```

##Task 4
Train the model for 15 epochs.


```python
history = model.fit(
    padded_sequences,
    np.array(labels),
    epochs=15,
    verbose=1
)
```

##Task 5
Predict the sentiment of any two new sentences of your choice and explain whether the predictions are reasonable.


```python
new_sentences = [
    "I love swimming",
    "This exam is horrible"
]

new_sequences = tokenizer.texts_to_sequences(new_sentences)

new_padded = pad_sequences(
    new_sequences,
    maxlen=max_length,
    padding="post"
)

predictions = model.predict(new_padded)

for sentence, pred in zip(new_sentences, predictions):
    sentiment = "Positive" if pred[0] >= 0.5 else "Negative"
    print(f"{sentence} --> {sentiment} ({pred[0]:.4f})")
```

predictions are reasonable because the words "love" and "coding" are associated with positive example, and "exam" and "horrible" are new
